## 1. Traer el código del proyecto

```python
import os
import subprocess
import sys

REPO = "https://github.com/samuelam1/GESTOR_DE_GASTOS_1"
CARPETA = "/content/gestor_de_gastos"  # ruta absoluta: no depende de dónde estés

if os.path.isdir(os.path.join(CARPETA, ".git")):
    # Ya estaba clonado (por ejemplo, ejecutaste esta celda antes): lo actualiza.
    subprocess.run(["git", "-C", CARPETA, "pull", "--ff-only"], check=True)
else:
    clonado = subprocess.run(["git", "clone", REPO, CARPETA])
    if clonado.returncode != 0:
        raise SystemExit(
            "No se pudo clonar el repositorio. Si dice 'could not read Username', "
            "el repositorio es PRIVADO: en GitHub ve a Settings > General > Danger Zone "
            "> Change visibility > Make public, y vuelve a ejecutar esta celda.")

os.chdir(CARPETA)
sys.path.insert(0, CARPETA)

if not os.path.isdir(os.path.join(CARPETA, "src")):
    raise SystemExit(f"No encuentro la carpeta 'src' en {CARPETA}. Contenido: {os.listdir(CARPETA)}")
print("Código listo en", CARPETA, "->", sorted(os.listdir(CARPETA)))

# Alternativa sin GitHub: sube el .zip al panel de archivos de Colab y usa
# !unzip -q proyecto.zip -d /content/gestor_de_gastos
```
